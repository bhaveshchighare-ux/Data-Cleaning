```python
import pandas as pd
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.impute import KNNImputer
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv(
    'WA_Fn-UseC_-Telco-Customer-Churn.csv'
)

df.head()
```

```python
print(df.shape)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()

print(df.shape)
```

```python
df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
)
```

```python
df.isnull().sum()
```

```python
numeric_columns = df.select_dtypes(
    include=['int64','float64']
).columns

imputer = KNNImputer(
    n_neighbors=5
)

df[numeric_columns] = imputer.fit_transform(
    df[numeric_columns]
)
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x=df['MonthlyCharges']
)

plt.show()
```

```python
Q1 = df['MonthlyCharges'].quantile(0.25)

Q3 = df['MonthlyCharges'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

df = df[
    (df['MonthlyCharges'] >= lower) &
    (df['MonthlyCharges'] <= upper)
]

print(df.shape)
```

```python
binary_columns = [
    'gender',
    'Partner',
    'Dependents',
    'PhoneService',
    'PaperlessBilling',
    'Churn'
]

le = LabelEncoder()

for col in binary_columns:
    df[col] = le.fit_transform(
        df[col]
    )
```

```python
df = pd.get_dummies(
    df,
    columns=[
        'MultipleLines',
        'InternetService',
        'OnlineSecurity',
        'OnlineBackup',
        'DeviceProtection',
        'TechSupport',
        'StreamingTV',
        'StreamingMovies',
        'Contract',
        'PaymentMethod'
    ],
    drop_first=True
)
```

```python
df['TotalCharges'] = np.log1p(
    df['TotalCharges']
)
```

```python
scaler = StandardScaler()

df[
    ['tenure',
     'MonthlyCharges',
     'TotalCharges']
] = scaler.fit_transform(
    df[
        ['tenure',
         'MonthlyCharges',
         'TotalCharges']
    ]
)
```

```python
df.head()
```

```python
df.to_csv(
    'preprocessed_telco_churn.csv',
    index=False
)

print(
    "Dataset Saved Successfully"
)
```
